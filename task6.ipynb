```python
def is_safe(node, color, colors, graph):
    for neighbor in graph[node]:
        if colors.get(neighbor) == color:
            return False
    return True
def graph_coloring_util(graph, colors, vertices):
    if not vertices:
        return True
    current_node = vertices[0]
    color = 1
    while True:
        if is_safe(current_node, color, colors, graph):
            colors[current_node] = color
            if graph_coloring_util(graph, colors, vertices[1:]):
                return True
            del colors[current_node]
        color += 1
def graph_coloring(graph):
    colors = {}
    vertices = list(graph.keys())

    if graph_coloring_util(graph, colors, vertices):

        color_groups = {}
        for node, col in colors.items():
            color_groups.setdefault(col, []).append(node)

        for col, nodes in sorted(color_groups.items()):
            formatted_nodes = "{" + ", ".join(sorted(nodes)) + "}"
            print(f"Colored vertices - {formatted_nodes} = color {col}")
    else:
        print("No solution found")
graph = {
    'A': [ 'B','C'],
    'B': ['A', 'D'],
    'C': ['A', 'D'],
    'D': ['B', 'C', 'E'],
    'E': ['D']
}
graph_coloring(graph)
```
