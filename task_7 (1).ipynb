```python
from collections import deque
class MonkeyBananaProblem:
    def __init__(self):
        self.initial_state = {
            'on_floor(Monkey)',
            'at(Monkey, Center)',
            'at(Box, Window)'
        }
        self.goal = {
            'has(Monkey, Banana)'
        }
        self.operators = {
            'Grasp(Banana)': {
                'preconds': {
                    'on_box(Monkey)',
                    'at(Box, Center)',
                    'at(Monkey, Center)'
                },
                'add': {
                    'has(Monkey, Banana)'
                },
                'del': set()
            },

            'Climb(Box)': {
                'preconds': {
                    'on_floor(Monkey)',
                    'at(Monkey, Center)',
                    'at(Box, Center)'
                },
                'add': {
                    'on_box(Monkey)'
                },
                'del': {
                    'on_floor(Monkey)'
                }
            },

            'Push(Box, Window, Center)': {
                'preconds': {
                    'on_floor(Monkey)',
                    'at(Monkey, Window)',
                    'at(Box, Window)'
                },
                'add': {
                    'at(Monkey, Center)',
                    'at(Box, Center)'
                },
                'del': {
                    'at(Monkey, Window)',
                    'at(Box, Window)'
                }
            },

            'Walk(Floor, Center, Window)': {
                'preconds': {
                    'on_floor(Monkey)',
                    'at(Monkey, Center)'
                },
                'add': {
                    'at(Monkey, Window)'
                },
                'del': {
                    'at(Monkey, Center)'
                }
            }
        }

    def is_goal(self, state):
        return self.goal.issubset(state)
    def apply_operator(self, state, operator):
        action = self.operators[operator]
        if not action['preconds'].issubset(state):
            return None
        new_state = state.copy()
        new_state -= action['del']
        new_state |= action['add']
        return new_state
    def get_applicable_actions(self, state):
        applicable = []
        for operator in self.operators:
            if self.operators[operator]['preconds'].issubset(state):
                applicable.append(operator)
        return applicable
    def bfs(self):
        queue = deque()
        queue.append((self.initial_state, []))
        visited = set()
        while queue:
            state, path = queue.popleft()
            state_key = frozenset(state)
            if state_key in visited:
                continue
            visited.add(state_key)
            if self.is_goal(state):
                return path, state
            actions = self.get_applicable_actions(state)
            for action in actions:
                new_state = self.apply_operator(state, action)
                if new_state is not None:
                    new_path = path + [action]
                    queue.append((new_state, new_path))
        return None, None
    def solve(self):
        path, final_state = self.bfs()
        if path is None:
            print("No solution found.")
            return
        print("MONKEY AND BANANA PROBLEM")
        print()
        print("Initial State:")
        for fact in sorted(self.initial_state):
            print(fact)
        print()
        print("Solution:")
        for i, action in enumerate(path, start=1):
            print(f"{i}. {action}")
        print()
        print("Final State:")
        for fact in sorted(final_state):
            print(fact)
        print()
        print("Goal reached successfully!")
if __name__ == "__main__":
    problem = MonkeyBananaProblem()
    problem.solve()
```
