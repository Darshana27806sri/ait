```python
def heuristic(state,goal):
    count=0
    for i in range(3):
        for j in range(3):
            if state[i][j]!=0 and state[i][j]!=goal[i][j]:
                count+=1
    return count
def print_puzzle(state):
    for row in state:
        print(row)
    print()
def hill_climbing(start,goal):
    current=start
    while True:
        print("Current State:")
        print_puzzle(current)
        current_h=heuristic(current,goal)
        if current_h==0:
            print("Goal State Reached!")
            return
        best=None
        best_h=current_h
        x,y=0,0
        for i in range(3):
            for j in range(3):
                if current[i][j]==0:
                    x,y=i,j
        moves=[(-1,0),(1,0),(0,-1),(0,1)]
        for dx,dy in moves:
            nx,ny=x+dx,y+dy
            if 0<=nx<3 and 0<=ny<3:
                neighbour=[row[:] for row in current]
                neighbour[x][y],neighbour[nx][ny]=neighbour[nx][ny],neighbour[x][y]
                h=heuristic(neighbour,goal)
                if h<best_h:
                    best_h=h
                    best=neighbour
        if best is None:
            print("No better state found.")
            print("Hill Climbing stopped.")
            return
        current=best
start=[[1,2,3],[4,0,6],[7,5,8]]
goal=[[1,2,3],[4,5,6],[7,8,0]]
print("Starting Puzzle:")
print_puzzle(start)
print("Goal Puzzle:")
print_puzzle(goal)
hill_climbing(start,goal)
```
