```python
from kanren import Relation, facts, var, run, conde
parent = Relation()
facts(
    parent,
    ("Abe", "Homer"),
    ("Homer", "Bart"),
    ("Homer", "Lisa")
)
def grandparent(x, z):
    y = var()
    return conde((parent(x, y), parent(y, z)))
x = var()
result = run(0, x, grandparent(x, "Bart"))
print("The grandparent of Bart is:", result)
```
