```python

R1 = 30
R2 = 70
total = R1 + R2
P1 = R1 / total
P2 = R2 / total
print("Path 1 Probability:", P1, "or", P1 * 100, "%")
print("Path 2 Probability:", P2, "or", P2 * 100, "%")
if P1 > P2:
    print("Most ants follow Path 1.")
elif P2 > P1:
    print("Most ants follow Path 2.")
else:
    print("Both paths have equal probability.")
print("\nResult:")
print(f"Path 2 has the higher probability of {P2 * 100:.0f}%.")
print("Therefore, most ants are expected to follow the minimum path.")
```
